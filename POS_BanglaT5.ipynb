```python
import os
import gc
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
from bnlp import BengaliPOS
import sacrebleu
from rouge_score import rouge_scorer
from nltk.translate.meteor_score import meteor_score
import nltk
nltk.download('wordnet')
import warnings
warnings.filterwarnings('ignore')

# ── FIXED SEED FOR REPRODUCIBILITY ──────────────────────────────────────────
torch.manual_seed(42)
torch.cuda.manual_seed(42)
np.random.seed(42)

# ── CONFIG ───────────────────────────────────────────────────────────────────
DATA_DIR = "/content/data"
MODEL_DIR = "/content/drive/MyDrive/dialect_nmt/models"
RESULTS_DIR = "/content/drive/MyDrive/dialect_nmt/results"
MODEL_NAME = "csebuetnlp/banglat5"
TARGET_COL = "bangla_speech"
MAX_LEN = 128
BATCH_SIZE = 8
EPOCHS = 50
LEARNING_RATE = 5e-5
PATIENCE = 5
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

DIALECTS = [
    {"name": "Chittagong",  "col": "chittagong_bangla_speech"},
    {"name": "Noakhali",    "col": "noakhali_bangla_speech"},
    {"name": "Sylhet",      "col": "sylhet_bangla_speech"},
    {"name": "Barishal",    "col": "barishal_bangla_speech"},
    {"name": "Mymensingh",  "col": "mymensingh_bangla_speech"},
]

POS_TAGS = ['NN','NNS','NNP','NNPS','VB','VBD','VBG','VBN','VBP','VBZ',
            'JJ','JJR','JJS','RB','RBR','RBS','PRP','PRP$','DT','IN',
            'CC','CD','UH','WP','OTHER']
POS2ID = {tag: idx for idx, tag in enumerate(POS_TAGS)}
NUM_POS_TAGS = len(POS_TAGS)

print(f"Using device: {DEVICE}")
print("Loading POS tagger and tokenizer...")
pos_tagger = BengaliPOS()
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
print("Ready.\n")

# ── POS HELPER ───────────────────────────────────────────────────────────────
def get_pos_ids(text, length):
    try:
        tokens_tags = pos_tagger.tag(text)
        tag_ids = [POS2ID.get(tag, POS2ID['OTHER']) for _, tag in tokens_tags]
        if len(tag_ids) < length:
            tag_ids = tag_ids + [POS2ID['OTHER']] * (length - len(tag_ids))
        else:
            tag_ids = tag_ids[:length]
    except:
        tag_ids = [POS2ID['OTHER']] * length
    return tag_ids

# ── DATASET ──────────────────────────────────────────────────────────────────
class DialectDataset(Dataset):
    def __init__(self, df, dialect_col):
        self.sources = df[dialect_col].tolist()
        self.targets = df[TARGET_COL].tolist()

    def __len__(self):
        return len(self.sources)

    def __getitem__(self, idx):
        source = self.sources[idx]
        target = self.targets[idx]
        model_inputs = tokenizer(source, max_length=MAX_LEN, padding='max_length', truncation=True, return_tensors='pt')
        labels = tokenizer(target, max_length=MAX_LEN, padding='max_length', truncation=True, return_tensors='pt')
        input_ids = model_inputs['input_ids'].squeeze()
        attention_mask = model_inputs['attention_mask'].squeeze()
        label_ids = labels['input_ids'].squeeze()
        label_ids[label_ids == tokenizer.pad_token_id] = -100
        pos_ids = torch.tensor(get_pos_ids(source, MAX_LEN), dtype=torch.long)
        return {'input_ids': input_ids, 'attention_mask': attention_mask, 'labels': label_ids, 'pos_ids': pos_ids}

# ── MODEL ────────────────────────────────────────────────────────────────────
class POSAugmentedBanglaT5(nn.Module):
    def __init__(self):
        super().__init__()
        self.model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME)
        embed_dim = self.model.config.d_model
        self.pos_embedding = nn.Embedding(NUM_POS_TAGS, embed_dim)
        self.pos_proj = nn.Linear(embed_dim, embed_dim)

    def forward(self, input_ids, attention_mask, labels, pos_ids):
        encoder = self.model.encoder
        token_embeds = encoder.embed_tokens(input_ids)
        pos_embeds = self.pos_proj(self.pos_embedding(pos_ids))
        combined_embeds = token_embeds + pos_embeds
        encoder_outputs = encoder(inputs_embeds=combined_embeds, attention_mask=attention_mask)
        outputs = self.model(attention_mask=attention_mask, encoder_outputs=encoder_outputs, labels=labels)
        return outputs.loss

    def generate(self, input_ids, attention_mask, pos_ids, **kwargs):
        encoder = self.model.encoder
        token_embeds = encoder.embed_tokens(input_ids)
        pos_embeds = self.pos_proj(self.pos_embedding(pos_ids))
        combined_embeds = token_embeds + pos_embeds
        encoder_outputs = encoder(inputs_embeds=combined_embeds, attention_mask=attention_mask)
        return self.model.generate(attention_mask=attention_mask, encoder_outputs=encoder_outputs, **kwargs)

# ── METRICS ──────────────────────────────────────────────────────────────────
def wer(ref, hyp):
    r, h = ref.split(), hyp.split()
    if len(r) == 0: return 0
    d = np.zeros((len(r)+1, len(h)+1))
    for i in range(len(r)+1): d[i][0] = i
    for j in range(len(h)+1): d[0][j] = j
    for i in range(1, len(r)+1):
        for j in range(1, len(h)+1):
            d[i][j] = d[i-1][j-1] if r[i-1]==h[j-1] else 1+min(d[i-1][j], d[i][j-1], d[i-1][j-1])
    return d[len(r)][len(h)] / len(r)

def cer(ref, hyp):
    if len(ref) == 0: return 0
    d = np.zeros((len(ref)+1, len(hyp)+1))
    for i in range(len(ref)+1): d[i][0] = i
    for j in range(len(hyp)+1): d[0][j] = j
    for i in range(1, len(ref)+1):
        for j in range(1, len(hyp)+1):
            d[i][j] = d[i-1][j-1] if ref[i-1]==hyp[j-1] else 1+min(d[i-1][j], d[i][j-1], d[i-1][j-1])
    return d[len(ref)][len(hyp)] / len(ref)

# ── ROUGE FIX (word mapping for Bangla) ──────────────────────────────────────
def compute_rouge(predictions, references):
    word_to_id = {}
    def encode(text):
        words = []
        for w in text.split():
            if w not in word_to_id:
                word_to_id[w] = f"w{len(word_to_id)}"
            words.append(word_to_id[w])
        return " ".join(words)

    rouge = rouge_scorer.RougeScorer(['rouge1','rouge2','rougeL'], use_stemmer=False)
    r1, r2, rl = [], [], []
    for pred, ref in zip(predictions, references):
        if pred.strip() == '' or ref.strip() == '':
            r1.append(0.0); r2.append(0.0); rl.append(0.0)
        else:
            scores = rouge.score(encode(ref), encode(pred))
            r1.append(scores['rouge1'].fmeasure)
            r2.append(scores['rouge2'].fmeasure)
            rl.append(scores['rougeL'].fmeasure)
    return np.mean(r1), np.mean(r2), np.mean(rl)

# ── MAIN LOOP ────────────────────────────────────────────────────────────────
all_results = []

for dialect_info in DIALECTS:
    DIALECT = dialect_info["name"]
    DIALECT_COL = dialect_info["col"]

    print(f"\n{'='*60}")
    print(f"TRAINING: {DIALECT}")
    print(f"{'='*60}")

    # reset seed per dialect for fair comparison
    torch.manual_seed(42)
    torch.cuda.manual_seed(42)
    np.random.seed(42)

    def load_data(split):
        path = os.path.join(DATA_DIR, f"{DIALECT} {split} Translation.csv")
        df = pd.read_csv(path)
        df.columns = df.columns.str.strip()
        df = df[[DIALECT_COL, TARGET_COL]].dropna()
        df[DIALECT_COL] = df[DIALECT_COL].astype(str).str.strip()
        df[TARGET_COL] = df[TARGET_COL].astype(str).str.strip()
        return df

    train_df = load_data("Train")
    val_df = load_data("Validation")
    test_df = load_data("Test")
    print(f"Train: {len(train_df)}, Val: {len(val_df)}, Test: {len(test_df)}")

    train_loader = DataLoader(DialectDataset(train_df, DIALECT_COL), batch_size=BATCH_SIZE, shuffle=True)
    val_loader = DataLoader(DialectDataset(val_df, DIALECT_COL), batch_size=BATCH_SIZE)
    test_loader = DataLoader(DialectDataset(test_df, DIALECT_COL), batch_size=BATCH_SIZE)

    model = POSAugmentedBanglaT5().to(DEVICE)
    optimizer = torch.optim.AdamW(model.parameters(), lr=LEARNING_RATE)

    best_val_loss = float('inf')
    patience_counter = 0
    best_model_path = os.path.join(MODEL_DIR, f"{DIALECT}_best_model.pt")

    # 1. Create lists to track the loss for your graph!
    history_train_loss = []
    history_val_loss = []

    for epoch in range(EPOCHS):
        model.train()
        total_loss = 0
        for batch in train_loader:
            input_ids = batch['input_ids'].to(DEVICE)
            attention_mask = batch['attention_mask'].to(DEVICE)
            labels = batch['labels'].to(DEVICE)
            pos_ids = batch['pos_ids'].to(DEVICE)

            optimizer.zero_grad()
            loss = model(input_ids, attention_mask, labels, pos_ids)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            total_loss += loss.item()

        avg_train_loss = total_loss / len(train_loader)

        model.eval()
        val_loss = 0
        with torch.no_grad():
            for batch in val_loader:
                input_ids = batch['input_ids'].to(DEVICE)
                attention_mask = batch['attention_mask'].to(DEVICE)
                labels = batch['labels'].to(DEVICE)
                pos_ids = batch['pos_ids'].to(DEVICE)

                loss = model(input_ids, attention_mask, labels, pos_ids)
                val_loss += loss.item()

        avg_val_loss = val_loss / len(val_loader)

        # 2. Save the numbers for this epoch
        history_train_loss.append(avg_train_loss)
        history_val_loss.append(avg_val_loss)

        print(f"Epoch {epoch+1}/{EPOCHS} | Train: {avg_train_loss:.4f} | Val: {avg_val_loss:.4f}")

        if avg_val_loss < best_val_loss:
            best_val_loss = avg_val_loss
            patience_counter = 0
            torch.save(model.state_dict(), best_model_path)
            print(f"  -> Best model saved")
        else:
            patience_counter += 1
            print(f"  -> No improvement ({patience_counter}/{PATIENCE})")
            if patience_counter >= PATIENCE:
                print(f"Early stopping at epoch {epoch+1}")
                break

    # 3. DRAW AND SAVE THE GRAPH IMMEDIATELY AFTER TRAINING!
    import matplotlib.pyplot as plt
    plt.figure(figsize=(8, 5))
    epochs_range = range(1, len(history_train_loss) + 1)
    plt.plot(epochs_range, history_train_loss, label="Training Loss", color="#1f77b4", linewidth=2)
    plt.plot(epochs_range, history_val_loss, label="Validation Loss", color="#ff7f0e", linestyle="--", linewidth=2)
    plt.xlabel("Epochs", fontweight='bold', fontsize=12)
    plt.ylabel("Loss", fontweight='bold', fontsize=12)
    plt.title(f"{DIALECT} Training Convergence", fontweight='bold', fontsize=14)
    plt.legend(fontsize=11)
    plt.grid(True, linestyle='--', alpha=0.7)

    graph_path = os.path.join(RESULTS_DIR, f"{DIALECT}_loss_curve.png")
    plt.savefig(graph_path, dpi=300, bbox_inches='tight')
    plt.show()
    print(f"Graph safely stored in your Drive at: {graph_path}")

    # ── EVALUATION ───────────────────────────────────────────────────────────
    print(f"\nEvaluating {DIALECT}...")
    model.load_state_dict(torch.load(best_model_path))
    model.eval()
    predictions, references = [], []

    with torch.no_grad():
        for batch in test_loader:
            input_ids = batch['input_ids'].to(DEVICE)
            attention_mask = batch['attention_mask'].to(DEVICE)
            pos_ids = batch['pos_ids'].to(DEVICE)
            generated = model.generate(
                input_ids=input_ids, attention_mask=attention_mask, pos_ids=pos_ids,
                max_length=MAX_LEN, num_beams=4, early_stopping=True
            )
            decoded_preds = tokenizer.batch_decode(generated, skip_special_tokens=True)
            decoded_labels = tokenizer.batch_decode(
                batch['labels'].masked_fill(batch['labels'] == -100, 0),
                skip_special_tokens=True
            )
            predictions.extend(decoded_preds)
            references.extend(decoded_labels)

    bleu = sacrebleu.corpus_bleu(predictions, [references]).score
    r1, r2, rl = compute_rouge(predictions, references)
    met = np.mean([meteor_score([r.split()], p.split()) if p.strip() and r.strip() else 0.0
                   for p, r in zip(predictions, references)])
    wer_score = np.mean([wer(r, p) for r, p in zip(references, predictions)])
    cer_score = np.mean([cer(r, p) for r, p in zip(references, predictions)])

    print(f"\n{DIALECT} RESULTS:")
    print(f"BLEU:    {bleu:.2f}")
    print(f"ROUGE-1: {r1:.4f} | ROUGE-2: {r2:.4f} | ROUGE-L: {rl:.4f}")
    print(f"METEOR:  {met:.4f}")
    print(f"WER:     {wer_score:.4f} | CER: {cer_score:.4f}")

    result = {
        'dialect': DIALECT, 'bleu': bleu, 'rouge1': r1, 'rouge2': r2,
        'rougeL': rl, 'meteor': met, 'wer': wer_score, 'cer': cer_score
    }
    all_results.append(result)
    pd.DataFrame([result]).to_csv(
        os.path.join(RESULTS_DIR, f"{DIALECT}_results.csv"), index=False
    )
    print(f"Saved {DIALECT}_results.csv")

    del model, optimizer, train_loader, val_loader, test_loader
    torch.cuda.empty_cache()
    gc.collect()
    print(f"GPU memory cleared. Moving to next dialect...\n")

# ── FINAL SUMMARY TABLE ──────────────────────────────────────────────────────
print("\n" + "="*60)
print("ALL DIALECTS COMPLETE — FINAL RESULTS")
print("="*60)
final_df = pd.DataFrame(all_results)
final_df.to_csv(os.path.join(RESULTS_DIR, "ALL_RESULTS.csv"), index=False)
print(final_df.to_string(index=False))
print("\nSaved ALL_RESULTS.csv to Drive.")
```
